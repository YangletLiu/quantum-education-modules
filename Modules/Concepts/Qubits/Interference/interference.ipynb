# Interference

## 1. Interference in Classical Physics


### 1.1 Waves
Interference is a phenomenon that occurs with waves. Waves are a crucial part of classical physics, and they are defined as distrubances that travel through a medium. For example, sound travels as a wave with air as its medium, and ripples in a pond are waves that use water as their medium. 

Waves have high points called crests and low points called troughs.

<img src="interference_images/wave_diagram.jpg"  width="400">


When waves interact, they combine to either amplify or cancel out, and this is called interference.


### 1.2 Types of Interference

**Constructive Interference**: When waves amplify each other, it is called constructive interference. This occurs when two crests or two troughs overlap. In the example below, the crests of the red and blue waves are perfectly aligned, (the troughs are also aligned and amplify each other) and they result in the black wave.

<img src="interference_images/constructive_interference.png"  width="400">

**Destructive Interference**: Destructive interference is when two waves cancel each other out. This occurs when a crest overlaps with a trough. In the following image, the two interacting waves cancel each other out, resulting in the flat black line in between them.

<img src="interference_images/destructive_interference.png"  width="400">

Note that it is rare for waves to be perfectly constructive or destructive. It is more likely that waves partially overlap, causing both constructive interference and destructive interference. An example is below.

<img src="interference_images/waves_not_aligned.png"  width="400">



When waves interact, they can result in a different wave completely. For example, when sound waves interfere, the listener may hear a different frequency or volume. Understanding interference is crucial to seeing how waves interact with each other and how they are received.

## 2. Interference in Quantum

### 2.1 Wavelike Behavior of Particles

In a quantum setting, interference is slightly different. Instead of looking at physical waves, quantum looks at the wave like behavior of particles, which are represented by **wave functions**. Wave functions signify the possible paths a particle could take. The probability of each path being taken is called the **probability amplitude**. When two wavefunctions interact, their probabilty amplitudes either interfere constructively or destructively, which is analogous to how amplitudes amplify or cancel in a classical setting.

When the waves interact constructively, the odds of the particle taking a particular path increase. When they interact destructively, it is less likely for a particle to take that path.

An important thing to note is that the concept of a wave function only exists when a particle is in superposition. The moment the state of a particle is measured, the wave function collapses. 

### 2.2 Quantum Interference with Qubits

Interference plays a large role in quantum computing because the wavefuctions of qubits affect each other. Here, interference represents possible computational paths. Sometimes these paths can cancel out, while others can reinforce a specific path. This means that a quantum computer does not compute one path at a time like a classical computer; rather, it can look at all possible computational paths at once. 

## 3. Young's Double Slit Experiment

Famously, physicist Thomas Young conducted his Double-Slit Experiment that perfectly demonstrates wave interference patterns.

The set up for the experiment is simple. Young had a metal plate with two slits in it, and behind it, he had a source of light. On the other side of the metal, there was a screen where he could see the pattern made by the light after passing through the slits. 

If light acted like a particle, then the resulting pattern on the wall should have been two bands of light. As demonstrated in the photo below.

<img src="interference_images/double_slit_particle.jpg"  width="400">



However, this is not what was observed, because light can also act like a wave.

The pattern created by the light was slivers of light broken up by slits of blackness. This is because, when the light travelled through the two slits, it split into two waves, and those two waves interacted with each other causing constructive interference and destructive interference. The constructive interference caused the slits of bright light, and the destructive interference caused the slits of black. Furthermore, the light pattern gets fainter the futher it is from the center.
