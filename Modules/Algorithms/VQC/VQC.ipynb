# Variational Quantum Classifier Algorithm

The Variational Quantum Classifer (VQC) is a hybrid classical-quantum algorithm that is used in Quantum Machine Learning (QML). It uses quantum circuits and pairs it with classical optimization. The purpose of the algorithm is to recognize patterns between data points by looking at their features. It is increasingly being used in statistical and data analysis in a variety of fields.

## 1. Overview

### 1.1 What is VQC

As previously stated, VQC takes the best of both classical computing and quantum computing. It takes classical data encodes it into quantum states, and processes them through a quantum circuit called a variational circuit. These circuits manipulate the data in a high-dimentional space to find accurate classfications. Then a classical optimizer is used to minimize error.  

In modern day, VQCs are being used for:
- image classification
- natural language processing
- fraud detection
- drug discovery
- financial modeling


### 1.2 Why VQC

VQC is part of a set of algorithms that can be run on Noisy Intermediate-Scale Quantum (NISQ) devices. Essentially this means is that VQC is an algorithm that gives meaningful results on quantum computers in their current state unlike theoretical algorithms like Shor's. Outside of this, it is very versatile and can be used in a variety of fields. 


## 2. VQC Algorithm Steps:

1) Data preparation (Preprocessing)

    The first step for running VQC on a dataset is to clean the data and to put it into an acceptable format. The features of the data are normalized and scaled so that they can be encoded into quantum states. Generally, the number of features should match the number of qubits, so part of this preprocessing is keeping the most important features and removing the rest.

2) Data Encoding (Feature Mapping)

    Data encoding is the process where all the data fed into the algorithm is converted into a quantum state. This is done using a feature map, where is data point has quantum gates applied on it. The result is a qunatum representation of the data in a high-dimentional Hilbert space. The classical information is encoded into qubits.

3) Initialize Parameterized Quantum Circuit(Ansatz)

    Then an initial adjustable circuit is created with adjustable parameters. The structure of this circuit is the ansatz. The ansatz has layers of quantum gates that rotate data points in a higher field.

4) Apply the Quantum Circuit

    The encoded data is passed through the circuit. The ansatz performs the transformation, and entanglement allows for qubits to interact with each other, forming more nuanced connections between qubits and features. 

5) Measurement

    Then the quantum state is measured and turned into probabilities. This is how we can extract usable prediction from the quantum system. The probabilites are mapped into class labels. 

6) Compute Loss Function

    Then the actual classification is compared to the estimation done by the VQC. From this, a loss function is calculated to see how accurate the algorithm was with its prediction. 

7) Classical Optimization

    Now, this is where classical techniques are used. Classical optimization techniques are used like gradient descent, COBYLA, SPSA are used to adjust the parameters to minimize the loss function. 

8) Iteration

    Steps 4-7 are repeated multiple times with updated parameters to see what the best classification is. 

9) Model Convergence

    The training stops when the loss function stabilizes or when it is manually stopped.

## 3. Optimization Techniques

There are many techniques that can be used to optimize the loss function. Below a few are listed. Optimization plays an important role in VQC. This is how we can find the best out of all the classifications that are generated.

### 3.1 Gradient Based Methods

Gradient based optimization methods use the gradient (or the slope) of the loss function to update the parameters. To calculate gradient, we change the parameter slightly upwards and slightly downwards and then estimate the slope. 

1. Gradient Descent

    This is the most common type of optimization. It simply moves the parameters in the direction that minimizes loss. While it is very simple, it can be very slow. 

2. Stochastic Gradient Descent (SGD)

    Instead of using the whole dataset, SGD updates parameters one data point at a time. This allows for faster updates, and introduces randomness. However, it is less stable as well. 

3. Natural Gradient

    Natural gradients are useful because they consider the geometry of quantum states and use the quantum Fisher information matrix for better convergence in most cases. However, it is very computationally expensive.

Some issues with gradient based optimization is that they are generally very computationally expensive and they are very sensitive to noise. They also lose accuracy in cases with very small gradients.

### 3.2 Non-Gradient Methods

There are also many optimization techniques that are gradient free.

1. Contrained Optimization By Linear Approximation (COBYLA)
    
    Instead of actually computing the gradient, it estimates the loss function's behavior at a point and uses that to change the parameters. It works veyr well with noisy systems, but it is difficult to scale.

2. Simultaneous Perturbation Stochastic Approximation (SPSA)
    
    This method randomly changes all the parameters at once and then estimates the loss change. Then it updates the parameters based on this loss. It is very efficient because it is testing all the data points at once and it can scale very well. The issue is that it is less precise and requires fine tuning.

The type of optimization that is used is dependent of the data set and what needs to be done. 

## 4. Circuit Design

There are two main factors that go into the circuit design for a VQC. The first is feature map and the second is the ansatz. Circuit design is what defines how dta is represented and processed inside the quantum model.

### 4.1 Feature Map

Feature maps encode classical data into qubits. And there are a few types of feature maps.

1) Angle Encoding

    Angle encoding is the most common and simple kind of feature map. Each data value is used as a rotaiton angle. The issue with using this method is that there is limit of how much can be expressed for more complex data. 

2) Amplitude Encoding

    In amplitude encoding, data is encoded into the amplitude of a quantum system. This works much better for large datasets with less qubits. It is very compact. However, it is very difficult to prepare and requires very complex qubits. 

3) Basis Encoding

    In basis encoding, data points are stores as either zeros or ones. While this is very intuitive and simple to execute, it is not efficient for all kinds of data. 

### 4.2 Ansatz

The ansatz is the part of the circuit that is trained. The ansatz uses rotation gates which is what the parameters control. It also has entangled gates. This is what allows for more nuanced relationships between features. There are also layers that allow for more flexibility.

There are two types of Ansatz:

1) Hardware-Efficient Ansatz

These are circuits that are built to work on modern quantum computers. Since quantum computers are still in development, these circuits tends to be very simple and have very few layers. They also struggle to capture complex relationships and really heavy loads of data.

2) Problem-Inspired Ansatz

These are circuits that are tailored to solve a particular kind of problem. The issue that arises is that these circuits may not be feasible to even run on an actual quantum computer. They also take a large amount of time to design. 


There are two main metrics for understanding how good a circuit is, and that is depth and expressibility. Shalllow circuits are circuits that do not have as many layers. They run faster, produce less noise, but they are less expressive. Deep circuits are slower and produce more noise, but they are more expressive.

Expressibility is how well a model is able to capture complex relationships.
